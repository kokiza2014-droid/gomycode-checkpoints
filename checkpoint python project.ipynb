```python
import math
from typing import Callable


class Calculatrice:
    def __init__(self):
        """
        Initialise le dictionnaire contenant les opérations mathématiques
        disponibles et les fonctions correspondantes.
        """
        self.operations: dict[str, Callable[[float, float], float]] = {
            "+": lambda nombre1, nombre2: nombre1 + nombre2,
            "-": lambda nombre1, nombre2: nombre1 - nombre2,
            "*": lambda nombre1, nombre2: nombre1 * nombre2,
            "/": self.division
        }

    @staticmethod
    def division(nombre1: float, nombre2: float) -> float:
        """Effectue une division en vérifiant la division par zéro."""
        if nombre2 == 0:
            raise ZeroDivisionError("La division par zéro est impossible.")

        return nombre1 / nombre2

    def add_operation(
        self,
        symbole: str,
        fonction: Callable[[float, float], float]
    ) -> None:
        """
        Ajoute une nouvelle opération au dictionnaire.

        Args:
            symbole: symbole représentant l'opération.
            fonction: fonction associée à cette opération.
        """
        if not isinstance(symbole, str):
            raise TypeError("Le symbole de l'opération doit être une chaîne.")

        if not callable(fonction):
            raise TypeError("La fonction fournie doit être appelable.")

        self.operations[symbole] = fonction

    def calculate(
        self,
        nombre1: float,
        symbole: str,
        nombre2: float
    ) -> float:
        """
        Effectue un calcul en utilisant l'opération correspondant au symbole.
        """
        if not isinstance(nombre1, (int, float)):
            message = "Erreur : le premier élément doit être un nombre."
            print(message)
            raise TypeError(message)

        if not isinstance(nombre2, (int, float)):
            message = "Erreur : le deuxième élément doit être un nombre."
            print(message)
            raise TypeError(message)

        if symbole not in self.operations:
            message = f"Erreur : l'opération '{symbole}' n'est pas valide."
            print(message)
            raise ValueError(message)

        try:
            fonction = self.operations[symbole]
            return fonction(nombre1, nombre2)

        except (ValueError, ZeroDivisionError) as erreur:
            print(f"Erreur de calcul : {erreur}")
            raise


# Fonctions mathématiques avancées

def exponentiation(nombre1: float, nombre2: float) -> float:
    """Calcule nombre1 élevé à la puissance nombre2."""
    return math.pow(nombre1, nombre2)


def racine_carree(nombre: float, nombre_inutile: float) -> float:
    """
    Calcule la racine carrée du premier nombre.

    Le deuxième argument est conservé pour respecter la structure
    de la méthode calculate().
    """
    if nombre < 0:
        raise ValueError(
            "La racine carrée d'un nombre négatif n'est pas définie "
            "dans les nombres réels."
        )

    return math.sqrt(nombre)


def logarithme(nombre: float, base: float) -> float:
    """Calcule le logarithme d'un nombre selon la base indiquée."""
    if nombre <= 0:
        raise ValueError(
            "Le nombre utilisé pour le logarithme doit être supérieur à zéro."
        )

    if base <= 0 or base == 1:
        raise ValueError(
            "La base du logarithme doit être positive et différente de 1."
        )

    return math.log(nombre, base)


def demander_nombre(message: str) -> float:
    """Demande un nombre à l'utilisateur et vérifie sa validité."""
    valeur = input(message)

    try:
        nombre = float(valeur)
    except ValueError as erreur:
        print("Erreur : veuillez saisir une valeur numérique.")
        raise TypeError("La valeur saisie n'est pas un nombre.") from erreur

    if not isinstance(nombre, (int, float)):
        raise TypeError("La valeur saisie n'est pas un nombre.")

    return nombre


def programme_principal():
    calculatrice = Calculatrice()

    # Ajout des opérations avancées
    calculatrice.add_operation("**", exponentiation)
    calculatrice.add_operation("sqrt", racine_carree)
    calculatrice.add_operation("log", logarithme)

    print("=== Calculatrice Python ===")
    print("Opérations disponibles : +, -, *, /, **, sqrt, log")
    print("Saisissez 'q' comme opération pour quitter.")

    while True:
        symbole = input("\nEntrez le symbole de l'opération : ").strip().lower()

        if symbole == "q":
            print("Fermeture de la calculatrice.")
            break

        try:
            if symbole == "sqrt":
                nombre1 = demander_nombre(
                    "Entrez le nombre dont vous voulez calculer la racine : "
                )

                # Le deuxième argument n'est pas utilisé par la racine carrée.
                nombre2 = 0

            elif symbole == "log":
                nombre1 = demander_nombre(
                    "Entrez le nombre dont vous voulez calculer le logarithme : "
                )
                nombre2 = demander_nombre(
                    "Entrez la base du logarithme : "
                )

            else:
                nombre1 = demander_nombre("Entrez le premier nombre : ")
                nombre2 = demander_nombre("Entrez le deuxième nombre : ")

            resultat = calculatrice.calculate(
                nombre1,
                symbole,
                nombre2
            )

            print(f"Résultat : {resultat}")

        except (TypeError, ValueError, ZeroDivisionError):
            print("Le calcul n'a pas pu être effectué.")


if __name__ == "__main__":
    programme_principal()
```

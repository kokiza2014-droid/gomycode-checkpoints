```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    # Déposer de l'argent
    def deposit(self, amount):
        self.balance += amount
        print(f"{amount} DZD déposés sur le compte de {self.owner}.")

    # Retirer de l'argent
    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            print(f"{amount} DZD retirés du compte de {self.owner}.")
        else:
            print(f"Retrait refusé : solde insuffisant pour {self.owner}.")

    # Consulter le solde
    def check_balance(self):
        return self.balance


# Création d'une instance appelée mon_compte
mon_compte = Account("Sihem")

# Transactions
mon_compte.deposit(500)
mon_compte.withdraw(150)

# Vérification du solde
print(f"Solde de {mon_compte.owner} : {mon_compte.check_balance()} DZD")

print("-" * 40)

# Création de plusieurs comptes
compte2 = Account("Mohamed", 1000)
compte3 = Account("Younes", 200)

# Transactions sur Mohamed
compte2.withdraw(300)
compte2.deposit(200)

# Transactions sur Younes
compte3.withdraw(250)  # Solde insuffisant
compte3.deposit(100)

# Affichage des soldes
print(f"Solde de {compte2.owner} : {compte2.check_balance()} DZD")
print(f"Solde de {compte3.owner} : {compte3.check_balance()} DZD")
```

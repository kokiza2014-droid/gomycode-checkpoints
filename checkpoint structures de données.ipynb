Question 6

Écrivez un programme Python pour créer un ensemble.

Exemples : {0, 1, 2, 3, 4}

Écrire un programme Python pour itérer sur des ensembles.

Écrire un programme Python pour ajouter des membres dans un ensemble et pour retirer des éléments d'un ensemble donné.

```python
#Question1
Esample= [2, 3, 6]
resultat=1
for element in Esample:
    resultat=resultat*element
print("le résultat de la question 1:",resultat)

#question2
tri=[(2, 5), (1, 2), (4, 4), (2, 3), (2, 1)]
tri.sort(key=lambda x:x[1])
print ("le résultat de la question 2:",tri)

#question3
d1 = {'a' : 100, 'b' : 200, 'c':300}

d2 = {'a' : 300, 'b' : 200, 'd':400}

# 2. On parcourt le deuxième dictionnaire
for cle, valeur in d2.items():
    if cle in d1:
        d1[cle] += valeur  # Si la clé existe déjà, on additionne
    else:
        d1[cle] = valeur   # Si elle n'existe pas (comme 'd'), on la crée

print("le résultat de la question 3 est:",d1)

#Question4
# On demande le nombre à l'utilisateur 
n=int(input("entrer votre note d'exam en entier"))


# On initialise un dictionnaire vide
dictionnaire_carres = {}

# Boucle de 1 à n (inclus)
for i in range(1, n + 1):
    dictionnaire_carres[i] = i * i

# On affiche le résultat
print("le résulta de la question 4:" ,dictionnaire_carres)

#Question 5

ma_liste = [('item1', '12.20'), ('item2', '15.10'), ('item3', '24.5')]

# Tri de la liste
# - key=lambda x: float(x[1]) : convertit le deuxième élément de chaque tuple en nombre flottant pour le tri
# - reverse=True : trie dans l'ordre décroissant (du plus grand au plus petit)
ma_liste.sort(key=lambda x: float(x[1]), reverse=True)
print("le résultat de la question 5:",ma_liste)

#Question 6
# Création d'un ensemble avec des accolades {}
mon_ensemble = {0, 1, 2, 3, 4}
print("Ensemble initial :", mon_ensemble)

ensemble_vide = set()


mon_ensemble.add(5)

# On utilise la méthode .update() pour ajouter plusieurs éléments d'un coup
mon_ensemble.update([6, 7])

print("Après ajouts (5, 6, 7) :", mon_ensemble)


#  ITÉRER SUR UN ENSEMBLE

print("Itération sur les éléments de l'ensemble:")
for element in mon_ensemble:
    print(f"Élément : {element}")
```

```python
def calculatrice(num1, num2):
    # Demander l'opérateur à l'utilisateur au sein de la fonction
    operateur = input("Entrez l'opérateur (+, -, *, /) : ")
    
    # Instructions conditionnelles pour déterminer l'opération
    if operateur == '+':
        return num1 + num2
    elif operateur == '-':
        return num1 - num2
    elif operateur == '*':
        return num1 * num2
    elif operateur == '/':
        # Gestion du cas de la division par zéro
        if num2 == 0:
            return "Erreur : Division par zéro impossible."
        return num1 / num2
    else:
        # Gestion du cas où l'opérateur n'est pas valide
        return "Erreur : Opérateur non valide."

# --- Fin de fonction ---

print("BIENVENUE DANS LA CALCULATRICE ")

try:
    # Demander à l'utilisateur d'entrer les deux nombres
    premier_nombre = float(input("Entrez le premier nombre : "))
    deuxieme_nombre = float(input("Entrez le deuxième nombre : "))
    
    # Appeler la fonction calculatrice et récupérer le résultat
    resultat = calculatrice(premier_nombre, deuxieme_nombre)
    
    # Imprimer le résultat
    print(f"Le résultat est : {resultat}")

except ValueError:
    # Sécurité si l'utilisateur saisit du texte au lieu d'un nombre
    print("Erreur : Veuillez entrer des nombres valides.")
```

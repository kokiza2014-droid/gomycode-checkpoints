```python
import numpy as np

# Création du tableau des notes
grades = np.array([85, 90, 88, 92, 95, 80, 75, 98, 89, 83])

# Calcul des statistiques
average = np.mean(grades)
median = np.median(grades)
std_dev = np.std(grades)

# Note maximale et minimale
max_grade = np.max(grades)
min_grade = np.min(grades)

# Indice de la note la plus élevée
max_index = np.argmax(grades)

# Nombre d'étudiants ayant une note supérieure à 90
count_above_90 = np.sum(grades > 90)

# Pourcentage d'étudiants ayant une note supérieure à 90
percentage_above_90 = np.mean(grades > 90) * 100

# Tableau des notes supérieures à 90
grades_above_90 = grades[grades > 90]

# Tableau des notes supérieures à 75
passing_grades = grades[grades > 75]

# Affichage des résultats
print("Tableau des notes :", grades)

print("\nStatistiques :")
print("Moyenne :", average)
print("Médiane :", median)
print("Écart type :", std_dev)

print("\nValeurs extrêmes :")
print("Note maximale :", max_grade)
print("Note minimale :", min_grade)

print("\nIndice de la note maximale :", max_index)

print("\nÉtudiants ayant une note supérieure à 90 :")
print("Nombre :", count_above_90)
print("Pourcentage :", percentage_above_90, "%")
print("Notes supérieures à 90 :", grades_above_90)

print("\nNotes supérieures à 75 :")
print(passing_grades)
```
